```python

!pip install -q transformers torch accelerate
from transformers import pipeline

generator = pipeline(
    "text-generation",
    model="TinyLlama/TinyLlama-1.1B-Chat-v1.0",
    max_new_tokens=180,
    do_sample=False
)
def generate_response(prompt):
    output = generator(prompt)
    return output[0]["generated_text"][len(prompt):].strip()

article = """
Artificial Intelligence (AI) is transforming many industries by enabling
machines to perform tasks that normally require human intelligence.
Machine Learning is a subset of AI that allows computers to learn from data.
Deep Learning, a further subset of Machine Learning, uses neural networks
with many layers to recognize complex patterns. AI is widely used in healthcare,
finance, education, transportation, and agriculture. Despite its advantages,
AI also raises concerns such as bias, privacy, security, and ethical issues.
Responsible AI development requires transparency, fairness, and accountability.
"""

prompt1 = f"""
Summarize the following article.

Article:
{article}
"""

print("="*70)
print("CYCLE 0 : BASIC PROMPT")
print("="*70)
print(generate_response(prompt1))

prompt2 = f"""
Summarize the following article.

Constraints:
1. Write between 70 and 90 words.
2. Use bullet points.
3. Include only the important ideas.

Article:
{article}
"""

print("\n" + "="*70)
print("CYCLE 1 : ADD WORD LIMIT + BULLET FORMAT")
print("="*70)
print(generate_response(prompt2))

prompt3 = f"""
Summarize the following article.

Constraints:
1. Write exactly 5 bullet points.
2. Total length should be about 80 words.
3. Mention:
   - Artificial Intelligence
   - Machine Learning
   - Deep Learning
   - Applications
   - Ethical challenges
4. Do not include unnecessary details.
5. Use simple language suitable for first-year engineering students.

Article:
{article}
"""

print("\n" + "="*70)
print("CYCLE 2 : REFINED PROMPT")
print("="*70)
print(generate_response(prompt3))


# =====================================================
# Evaluation
# =====================================================
print("\n" + "="*70)
print("EVALUATION")
print("="*70)

print("""
Cycle 0:
--------
• Generic summary.
• No control over length or structure.
• Important points may be omitted.

Cycle 1:
--------
• Word limit improves conciseness.
• Bullet format improves readability.
• Includes major ideas.

Cycle 2:
--------
• Exactly five bullet points.
• Covers all required topics.
• Simple language makes the summary easier to understand.
• Output is well-structured and follows all constraints.

Conclusion:
-----------
Constraint specification and iterative refinement significantly improve
the quality, consistency, and usability of LLM-generated responses.
""")


```
