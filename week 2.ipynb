```python
from transformers import pipeline
# Load a text generation model from Hugging Face
generator = pipeline(
    "text-generation",
    model="distilgpt2"
)
# Naive Prompt
naive_prompt = "Write a one-paragraph bio of Ada Lovelace."
# Enhanced Prompt
enhanced_prompt = """
You are a technology historian writing for a university newsletter.
Write a biography of Ada Lovelace.
Requirements:
1. Length: 120-150 words
2. Include:
   - Birth and death years
   - Collaboration with Charles Babbage
   - Contribution to the Analytical Engine
   - Why she is considered the first computer programmer
3. Tone: Professional and engaging
4. Format: Exactly one paragraph
"""
# Generate response
def get_response(prompt):
    output = generator(
        prompt,
        max_length=200,
        num_return_sequences=1,
        truncation=True
    )
    return output[0]["generated_text"]
# Execute prompts
naive_output = get_response(naive_prompt)
enhanced_output = get_response(enhanced_prompt)
# Print outputs
print("=" * 80)
print("NAIVE PROMPT OUTPUT")
print("=" * 80)
print(naive_output)
print("\n" + "=" * 80)
print("ENHANCED PROMPT OUTPUT")
print("=" * 80)
print(enhanced_output)
# Comparison
print("\n" + "=" * 80)
print("COMPARISON")
print("=" * 80)
print("""
1. Relevance:
   - Enhanced prompt is usually more focused on Ada Lovelace's achievements.
   - Naive prompt may contain less targeted information.
2. Completeness:
   - Enhanced prompt explicitly requests important facts.
   - Naive prompt may omit key details.
3. Style:
   - Enhanced prompt produces a professional and structured biography.
   - Naive prompt style depends entirely on the model.
Conclusion:
The enhanced prompt generally produces more relevant, complete, and well-structured output because it includes role framing, specificity, and formatting instructions.
""")

```
