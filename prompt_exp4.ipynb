```python
from transformers import pipeline
generator = pipeline(
    task="text-generation",
    model="TinyLlama/TinyLlama-1.1B-Chat-v1.0",
    return_full_text=False
)
def generate_response(prompt, max_tokens=150):
    output = generator(
        prompt,
        max_new_tokens=max_tokens,
        do_sample=True,
        temperature=0.7,
        top_p=0.9,
        repetition_penalty=1.1,
        clean_up_tokenization_spaces=False
    )
    return output[0]["generated_text"]
vague_prompt = """
### Instruction:
Write about Python.

### Response:
"""

print("=" * 60)
print("VAGUE PROMPT")
print("=" * 60)
print("Prompt:")
print("Write about Python.\n")

print("Generated Output:\n")
print(generate_response(vague_prompt, 100))
contradictory_prompt = """
### Instruction:
Write exactly 30 words.

Explain Machine Learning in detail with five examples.

### Response:
"""

print("\n" + "=" * 60)
print("CONTRADICTORY PROMPT")
print("=" * 60)
print("Prompt:")
print(contradictory_prompt)

print("\nGenerated Output:\n")
print(generate_response(contradictory_prompt, 150))
refined_prompt = """
### Instruction:

You are a computer science teacher.

Write one paragraph (120–150 words) explaining Machine Learning.

Requirements:
1. Use simple English.
2. Include these five applications:
   - Healthcare
   - Education
   - Banking
   - Agriculture
   - Transportation
3. End with a short conclusion.

### Response:
"""

print("\n" + "=" * 60)
print("REFINED PROMPT")
print("=" * 60)
print("Prompt:")
print(refined_prompt)

print("\nGenerated Output:\n")
print(generate_response(refined_prompt, 220))
```
