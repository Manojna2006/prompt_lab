```python
!pip install together

from dotenv import load_dotenv
import os
load_dotenv("hf.env")
print(os.getenv("HF"))

from huggingface_hub import InferenceClient
from dotenv import load_dotenv
import os
# ----------------------------------------
# Load Hugging Face Token
# ----------------------------------------
load_dotenv("hf.env")
HF_TOKEN = os.getenv("HF")
if HF_TOKEN is None:
    raise ValueError("HF not found in hf.env")
# ----------------------------------------
# Create Client
# ----------------------------------------
client = InferenceClient(
    api_key=HF_TOKEN
)

# ----------------------------------------
# Function to get model response
# ----------------------------------------
def get_response(prompt):

    completion = client.chat.completions.create(
        model="Qwen/Qwen2.5-7B-Instruct",   # Supported model
        messages=[
            {
                "role": "user",
                "content": prompt
            }
        ],
        max_tokens=200,
        temperature=0.7,
    )
    return completion.choices[0].message.content.strip()

# ----------------------------------------
# Prompt Iteration 1
# ----------------------------------------
prompt1 = """
Summarize the plot of the Shakespearean play Romeo and Juliet.
"""
prompt2 = """
Summarize the plot of the Shakespearean play Romeo and Juliet in exactly two sentences.
Use a clear, formal, and concise writing style.
"""
prompt3 = """
Summarize the plot of the Shakespearean play Romeo and Juliet in exactly two sentences.
Use a clear, formal, and concise style.
Include:
1. The setting (Verona, Italy)
2. The conflict between the Montague and Capulet families
3. The tragic ending
4. The central theme of love versus hatred
"""

# ----------------------------------------
# Generate Responses
# ----------------------------------------
output1 = get_response(prompt1)
output2 = get_response(prompt2)
output3 = get_response(prompt3)

# ----------------------------------------
# Print Results
# ----------------------------------------
print("=" * 80)
print("ITERATION 1: Minimal Instruction")
print("=" * 80)
print(output1)

print("\n" + "=" * 80)
print("ITERATION 2: Added Length and Style Constraints")
print("=" * 80)
print(output2)

print("\n" + "=" * 80)
print("ITERATION 3: Added Key Content Elements")
print("=" * 80)
print(output3)

# ----------------------------------------
# Improvement Analysis
# ----------------------------------------
print("\n" + "=" * 80)
print("DOCUMENTATION OF IMPROVEMENTS")
print("=" * 80)

print("""
Iteration 1:
- Uses only a minimal instruction.
- The response length and style may vary.
- Important details may be omitted.
Iteration 2:
- Adds an exact length requirement (two sentences).
- Requests a formal and concise writing style.
- Produces a more focused and consistent summary.
Iteration 3:
- Specifies the essential content to include:
    • Setting (Verona, Italy)
    • Family conflict
    • Tragic ending
    • Theme (Love versus Hatred)
- Produces the most informative and structured summary while satisfying all constraints.
""")
```
