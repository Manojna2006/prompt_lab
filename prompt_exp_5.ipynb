```python
from transformers import AutoTokenizer, AutoModelForSeq2SeqLM

# Load FLAN-T5 Base model
model_name = "google/flan-t5-base"

tokenizer = AutoTokenizer.from_pretrained(model_name)
model = AutoModelForSeq2SeqLM.from_pretrained(model_name)


# Function to generate answer
def generate_answer(prompt):
    inputs = tokenizer(prompt, return_tensors="pt")

    outputs = model.generate(
        **inputs,
        max_new_tokens=10
    )

    return tokenizer.decode(outputs[0], skip_special_tokens=True)


# -------------------------------
# ZERO-SHOT EXAMPLE
# -------------------------------

zero_prompt = """
Classify the sentiment as Positive, Negative, or Neutral.

Sentence:
I waited two hours for customer support, but they finally solved my issue.

Answer:
"""

print("========== ZERO-SHOT RESULT ==========")
print(generate_answer(zero_prompt))


# -------------------------------
# FEW-SHOT EXAMPLE
# -------------------------------

few_prompt = """
Classify the sentiment as Positive, Negative, or Neutral.

Example 1
Sentence: I love this phone.
Answer: Positive

Example 2
Sentence: The food was terrible.
Answer: Negative

Example 3
Sentence: The meeting starts at 10 AM.
Answer: Neutral

Now classify the following sentence.

Sentence:
I waited two hours for customer support, but they finally solved my issue.

Answer:
"""

print("\n========== FEW-SHOT RESULT ==========")
print(generate_answer(few_prompt))


# -------------------------------
# TEST SENTENCES
# -------------------------------

test_sentences = [
    "I absolutely love this laptop.",
    "The service was awful.",
    "The train leaves at 8 PM.",
    "The movie was boring but the ending was fantastic.",
    "The product arrived damaged."
]


# -------------------------------
# COMPARISON
# -------------------------------

print("\n" + "=" * 80)
print("ZERO-SHOT vs FEW-SHOT COMPARISON")
print("=" * 80)

for sentence in test_sentences:

    zero_prompt = f"""
Classify the sentiment as Positive, Negative, or Neutral.

Sentence:
{sentence}

Answer:
"""

    few_prompt = f"""
Classify the sentiment as Positive, Negative, or Neutral.

Example 1
Sentence: I love this phone.
Answer: Positive

Example 2
Sentence: The food was terrible.
Answer: Negative

Example 3
Sentence: The meeting starts at 10 AM.
Answer: Neutral

Now classify the following sentence.

Sentence:
{sentence}

Answer:
"""

    zero_result = generate_answer(zero_prompt)
    few_result = generate_answer(few_prompt)

    print("-" * 80)
    print(f"Sentence  : {sentence}")
    print(f"Zero-Shot : {zero_result}")
    print(f"Few-Shot  : {few_result}")
```
